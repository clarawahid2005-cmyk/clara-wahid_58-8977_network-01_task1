```python
categorical_columns = ["faculty", "city", "club", "fee_paid"]

for column in categorical_columns:
    print("\n" + "=" * 40)
    print(column.upper())
    print("=" * 40)
    counts = df[column].value_counts(dropna=False)
    for value, count in counts.items():
        print(f"{repr(value)} : {count}")
```

```python
for column in ["faculty", "club", "city"]:
    df[column] = df[column].str.strip().str.lower()
```

```python
print("FACULTY")
print(df["faculty"].value_counts())

print("\nCITY")
print(df["city"].value_counts())

print("\nCLUB")
print(df["club"].value_counts())
```

```python
faculty_map = {
    "met": "MET",
    "media engineering and technology": "MET",
    
    "iet": "IET",
    "information engineering and technology": "IET",
    
    "ems": "EMS",
    "engineering and materials science": "EMS",
    
    "pharmacy": "Pharmacy",
    "pharma": "Pharmacy",
    
    "management": "Management"
}

city_map = {
    "cairo": "Cairo",
    "giza": "Giza",
    "el giza": "Giza",
    "alex": "Alexandria",
    "alexandria": "Alexandria"
}

club_map = {
    "music": "Music",
    "robotics": "Robotics",
    "football": "Football",
    "soccer": "Football",
    "chess": "Chess",
    "chess club": "Chess",
    "debate": "Debate",
    "debate club": "Debate",
    "debating": "Debate"
}
```

```python
df["faculty"] = df["faculty"].map(faculty_map)
df["city"] = df["city"].map(city_map)
df["club"] = df["club"].map(club_map)
```

```python
print(df["faculty"].value_counts())
print()
print(df["city"].value_counts())
print()
print(df["club"].value_counts())
```

```python
assert set(df["faculty"].dropna().unique()) <= {
    "MET", "IET", "EMS", "Pharmacy", "Management"
}

assert set(df["city"].dropna().unique()) <= {
    "Cairo", "Giza", "Alexandria"
}

assert set(df["club"].dropna().unique()) <= {
    "Robotics", "Debate", "Music", "Football", "Chess"
}
```

```python
df["full_name"] = df["full_name"].str.strip()
df["full_name"] = df["full_name"].str.replace(r"\s+", " ", regex=True)
df["full_name"] = df["full_name"].str.title()
```

```python
df["email"] = df["email"].str.strip().str.lower()
```

```python
print(df[["full_name", "email"]].to_string(index=False))
```

```python
assert df["full_name"].equals(
    df["full_name"].str.strip().str.replace(r"\s+", " ", regex=True).str.title()
)

assert df["email"].equals(
    df["email"].str.strip().str.lower()
)
```

```python
df["fee_paid"].value_counts(dropna=False).to_string()
```

```python
fee_map = {
    "1": True,
    "yes": True,
    "y": True,
    "true": True,
    "0": False,
    "no": False,
    "n": False,
    "false": False
}
```

```python
df["fee_paid"] = (
    df["fee_paid"]
    .astype(str)
    .str.strip()
    .str.lower()
    .map(fee_map)
)
```

```python
print(df["fee_paid"].value_counts())
print(df["fee_paid"].dtype)
```

```python
assert df["fee_paid"].dtype == bool
assert set(df["fee_paid"].unique()) <= {True, False}
```
